```python
import pandas as pd
import requests
import time

# forms
# https://docs.google.com/forms/d/1HVHNYoUpInfg-0LW04_C17E9xguaCm3untFb5tV-_r8/edit#responses
# sheet
# https://docs.google.com/spreadsheets/d/1QqflzE7Ukzl_o9BcxudYZlL5yjV5B-7Qn-KLOCOo1wo/edit?resourcekey=&gid=1567690888#gid=1567690888

# Documentação da API:
# https://open-meteo.com/en/docs


endpoint_sheet = "https://docs.google.com/spreadsheets/d/1QqflzE7Ukzl_o9BcxudYZlL5yjV5B-7Qn-KLOCOo1wo/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
dados_planilha = pd.read_csv(endpoint_sheet)

print("---- Dados coletados ----")
display(dados_planilha)
```

```python
# iterar cada resposta do formulário
for index, linha in dados_planilha.iterrows():
  nome_local = linha["Nome do local"]
  lat = linha["Latitude"]
  lon = linha["Longitude"]

  parametros = {
      "latitude": lat,
      "longitude": lon,
      "current": "temperature_2m"
  }

  print(f"----- {nome_local} -----")
  resposta = requests.get(endpoint_api, params=parametros).json()
  display(resposta)
  print("\n")
```
